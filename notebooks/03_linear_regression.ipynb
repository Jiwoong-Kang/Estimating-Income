# 03. Linear Regression

회귀로 0/1 소득을 숫자로 예측한 뒤, 임계값(기본 0.5)으로 분류한다.
기준선: Dummy (`results/dummy.json`)

```python
import sys
import numpy as np
sys.path.append("..")

from sklearn.linear_model import LinearRegression
from sklearn.base import BaseEstimator, ClassifierMixin, clone
from src.data import load_adult
from src.preprocess import split_features_target, make_model_pipeline
from src.evaluate import cross_val_scores, evaluate_on_test, save_results, ThresholdingRegressor

X, y = split_features_target(load_adult("train"))
X_test, y_test = split_features_target(load_adult("test"))

print(X.shape, y.mean())
```

```python
model = ThresholdingRegressor(LinearRegression())
pipe = make_model_pipeline(model)

cv = cross_val_scores(pipe, X, y)
print("CV:", cv)

pipe2 = make_model_pipeline(clone(model))
test = evaluate_on_test(pipe2, X, y, X_test, y_test)
print("Test:", test["matrices"])
print("Confusion matrix:", test["confusion_matrix"])

path = save_results(
    "linear_regression",
    cv,
    test,
    extra={"threshold": 0.5},
)

```
